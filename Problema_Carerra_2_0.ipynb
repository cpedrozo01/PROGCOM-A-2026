```python
# ==========================================
# PLAN DE PRODUCCIÓN
# ==========================================

nombre_producto = "Silla ergonómica"

inventario_inicial = 20
stock_minimo = 5

demanda = [15, 20, 18, 25, 22, 30]

materiales = [
    ("Estructura metálica", 1),
    ("Asiento tapizado", 1),
    ("Ruedas", 5),
    ("Tornillos M6", 8),
]


# ------------------------------------------
# FUNCIÓN: calcular producción
# ------------------------------------------
def calcular_plan_produccion(demanda, inventario, stock_minimo):

    plan = []
    total_producido = 0

    for semana, pedido in enumerate(demanda, start=1):

        # Verificar si se necesita producir
        if inventario < pedido + stock_minimo:
            producir = (pedido + stock_minimo) - inventario
        else:
            producir = 0

        # Actualizar inventario
        inventario = inventario + producir - pedido

        # Guardar datos
        plan.append({
            "semana": semana,
            "demanda": pedido,
            "producir": producir,
            "inventario_final": inventario
        })

        total_producido += producir

    return plan, total_producido, inventario


# ------------------------------------------
# FUNCIÓN: mostrar tabla
# ------------------------------------------
def mostrar_plan(plan):

    print("=" * 65)
    print(f"{'PLAN DE PRODUCCIÓN':^65}")
    print("=" * 65)

    encabezado = (
        f"{'Semana':<10}"
        f"{'Demanda':<12}"
        f"{'Producción':<15}"
        f"{'Inventario Final':<18}"
    )

    print(encabezado)
    print("-" * 65)

    for fila in plan:
        print(
            f"{fila['semana']:<10}"
            f"{fila['demanda']:<12}"
            f"{fila['producir']:<15}"
            f"{fila['inventario_final']:<18}"
        )

    print("=" * 65)


# ------------------------------------------
# FUNCIÓN: calcular materiales
# ------------------------------------------
def calcular_materiales(materiales, total_producido):

    print("\n")
    print("=" * 65)
    print(f"{'MATERIALES NECESARIOS':^65}")
    print("=" * 65)

    print(f"Total a producir: {total_producido} unidades\n")

    for nombre, cantidad in materiales:

        total_material = cantidad * total_producido

        print(f"{nombre:<25} : {total_material} unidades")

    print("=" * 65)


# ==========================================
# PROGRAMA PRINCIPAL
# ==========================================

print("\n")
print(f"PRODUCTO: {nombre_producto}")
print("\n")

plan, total_producido, inventario_final = calcular_plan_produccion(
    demanda,
    inventario_inicial,
    stock_minimo
)

mostrar_plan(plan)

print(f"\nInventario final: {inventario_final} unidades")

calcular_materiales(materiales, total_producido)
```
